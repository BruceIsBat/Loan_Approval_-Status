```python
# import needed library
import pandas as pd
import numpy as np
from datetime import datetime
import matplotlib.pyplot as plt
```

```python
df_copy["id"] = df_copy["id"].astype(object)
df_copy["member_id"] = df_copy["member_id"].astype(object)
```

```python
df_copy = df_copy.rename(columns={'term': 'term_in_months'})
df_copy['term_in_months'] = df_copy['term_in_months'].str.replace(' months', '', regex=False).astype(int)
print(df_copy['term_in_months'].head())
print(df_copy['term_in_months'].dtype)
```

```python
df_copy['loan_amount'] = df_copy['loan_amount'].astype(float)
df_copy['total_payment'] = df_copy['total_payment'].astype(float)
```
