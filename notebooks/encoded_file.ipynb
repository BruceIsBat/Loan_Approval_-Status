```python
#1. load dataset
import pandas as pd

df = pd.read_csv("cleaned_data_after EDA (1).csv")
df.info()

#2. safe drop
if 'customer_id' in df.columns:
    df.drop(columns=['customer_id'], inplace=True)

#3.handling salary detected
df['salary_detected'] = df['salary_detected'].map({True:1, False: 0})


#4. handling states .....keep 10 states

top_states = df['state'].value_counts().index[:10]

df['state'] = df['state'].apply(lambda x: x if x in top_states else 'Other')

#5. main features that needs encoding
df = pd.get_dummies(
    df, columns=[
        'employment_type',
        'loan_purpose',
        'state'
    ], drop_first=True
)

#6. Seperate features and target

x = df.drop('default_90d',axis=1)
y = df['default_90d']


#7. train,test,split

from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=42,
    stratify=y
)


#8. scaling

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)


#9. save
import numpy as np
np.save('x_train.npy', x_train)
np.save('x_test.npy', x_test)
np.save('y_train.npy', y_train)
np.save('y_test.npy', y_test)


df.to_csv('cleaned and_encoded_data.csv', index=False)
```
