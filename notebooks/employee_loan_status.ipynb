# To Clean Data

* Removal of features not relevant to the prediction model

```python
# Removal of features that should not be used since they do not help predict creditworthiness for a new application.
df = df.drop("loan_id", axis=1) # Excluded because its an identifier, has no predictive relationship with loan default, and does not represent behavior or risk.
df = df.drop("customer_id", axis=1) # Excluded because its an identifier, has no predictive relationship with loan default, and does not represent behavior or risk.
df = df.drop("application_date", axis=1) # The exact date itself is rarely predictive, having nothing to do with loan payment being defaulted or not
df = df.drop("monthly_payment_ngn", axis=1) # This variable is derived from other variables. Including it introduces multicollinearity.
df = df.drop("disbursement_date", axis=1) # This happens after loan approval
df = df.drop("loan_status", axis=1) # Not a pure risk outcome.
df = df.drop("dpd_current", axis=1) # Days Past Due happens after loan repayment begins.
df = df.drop("dpd_max_90d", axis=1) # Same as reasons for dropping dpd_current. It is post-loan behavior.
df = df.drop("default_180d", axis=1) # Happens much later than dpd_max_180d.
```

# Check for Outliers

```python
df.describe()
```

The steps below is to automatically detect and treat outliers in all numeric columns in the cleaned dataset using the IQR (Interquartile Range) method

```python
# Automatically Detect Numeric Columns
numeric_cols = df.select_dtypes(include=['int64','float64']).columns

print("Numeric Columns:")
print(numeric_cols)
```

```python
# Detect Outliers in All Numeric Columns
outlier_summary = {}

for col in numeric_cols:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower) | (df[col] > upper)]

    outlier_summary[col] = len(outliers)

print("Outlier Count Per Column")
print(outlier_summary)
```

```python
# Automatically Treat Outliers (Capping Method)
for col in numeric_cols:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    df[col] = df[col].clip(lower, upper)
```

```python
# To verify That Outliers Are Removed
check_outliers = {}

for col in numeric_cols:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower) | (df[col] > upper)]

    check_outliers[col] = len(outliers)

print("Remaining Outliers After Treatment")
print(check_outliers)
```

* Handling states .....keep 10 states

```python
top_states = df['state'].value_counts().index[:10]
df['state'] = df['state'].apply(lambda x: x if x in top_states else 'Other')
```

* Main features that needs encoding

```python
df = pd.get_dummies(
    df, columns=[
        'employment_type',
        'loan_purpose',
        'state'
    ], drop_first=True
)
```

* Seperate features and target

```python
x = df.drop('default_90d',axis=1)
y = df['default_90d']
```

*  Train,test,split

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=42,
    stratify=y
)
print('Data split into Training (80%) and Testing (20%) sets.')
```

* Scaling

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)
```

* Save

```python
np.save('x_train.npy', x_train)
np.save('x_test.npy', x_test)
np.save('y_train.npy', y_train)
np.save('y_test.npy', y_test)

# df.to_csv('cleaned and_encoded_data.csv', index=False)
```
