```python
%%writefile app.py
import streamlit as st
import joblib
from PIL import Image

# Loading Model
model = joblib.load('/content/drive/MyDrive/credit_model.pkl')

def run():
  st.title('Credit Risk Assessment')

  Amount = st.number_input('principal_ngn')
  interest = st.number_input('interest_rate_annual')
  Duration_in_months = st.number_input('tenor_months')
  Annual_income = st.number_input('salary_ngn')
  DTI = st.number_input('debt_to_income_ratio')
  Credit_score = st.number_input('credit_score')

  Employment_list = ['private_sector', 'self_employed', 'unemployed']
  Employment_options =  list(range(len(Employment_list)))
  Employment = st.selectbox('Employment Status', Employment_options, format_func=lambda x: Employment_list[x])

  Loan_purpose = ['debt_consolidation', 'education', 'home_improvement', 'medical', 'personal']
  Loan_options = list(range(len(Loan_purpose)))
  Purpose = st.selectbox('Loan Purpose', Loan_options, format_func=lambda x: Loan_purpose[x])

  Salary_validation = ['Yes', 'No']
  Salary_options = list(range(len(Salary_validation)))
  Salary_detected = st.selectbox('Salary Detected', Salary_options, format_func=lambda x: Salary_validation[x])

  State_in_Nigeria = ['Anambra', 'Delta', 'Imo', 'Kaduna', 'Kano', 'Lagos', 'Ogun', 'Other', 'Oyo', 'Rivers']
  State_options = list(range(len(State_in_Nigeria)))
  State = st.selectbox('Location', State_options, format_func=lambda x: State_in_Nigeria[x])

  if st.button("PREDICT"):
    features =  [[Amount, interest, Duration_in_months, Annual_income, DTI, Credit_score, Employment, Purpose, State, Salary_detected ]]
    print(features)
    prediction = model.predict(features)
    output = round(prediction[0], 4)
    if output > 0.5: # Assuming 0.5 is our threshold
          st.success(f'Customer is likely to pay. Probability: {output}')
    else:
          st.error(f'High Risk. Probability: {output}')

if __name__ == '__main__':
  run()
```
